## **Imputation**

```python
print(f'  cores   : {df["cores"].isna().sum()}')
print(f'  threads : {df["threads"].isna().sum()}')
```

If the column value is NaN:
* Find the value from another laptop with the same processor.
* If there is no reference, fill in the string 'Unknown'.

If a value already exists, return it as is.

```python
proc_cores_map   = df.dropna(subset=['cores']).groupby('processor')['cores'].first()
proc_threads_map = df.dropna(subset=['threads']).groupby('processor')['threads'].first()

def fill_from_processor(row, col, lookup_map):
    if pd.isna(row[col]):
        return lookup_map.get(row['processor'], 'Unknown')
    return row[col]

df['cores']   = df.apply(lambda row: fill_from_processor(row, 'cores',   proc_cores_map),   axis=1)
df['threads'] = df.apply(lambda row: fill_from_processor(row, 'threads', proc_threads_map), axis=1)

print('Number of NaNs after filling:')
print(f'  cores   : {df["cores"].isna().sum()}')
print(f'  threads : {df["threads"].isna().sum()}')
print()
print('Number of rows filled with \'Unknown\' (no processor reference):')
print(f'  cores   : {(df["cores"] == "Unknown").sum()}')
print(f'  threads : {(df["threads"] == "Unknown").sum()}')
```

## **Make new Features**

### **1. Categorize Brand**

```python
print(df['name'].str.split().str[0].value_counts())
```

```python
known_brands = [
    'HP', 'Lenovo', 'Asus', 'MSI', 'Acer', 'Dell', 'Apple', 'Samsung',
    'Infinix', 'Microsoft', 'Gigabyte', 'Xiaomi', 'Thomson', 'Huawei',
    'Motorola', 'Honor', 'Ultimus', 'Alienware', 'Colorful', 'Chuwi',
    'Tecno', 'Zebronics', 'Avita', 'Razer', 'Red Magic', 'Air Falcon'
]

brand_mapping = {
    'ASUS' : 'Asus',
    'Red'  : 'Red Magic',
    'Air'  : 'Air Falcon',
}

# Remove hidden character U+200E (Left-to-Right Mark) from the entire 'name' column
df['name'] = df['name'].str.replace('\u200e', '', regex=False)

# Extract the first word as brand
df['brand'] = df['name'].str.split().str[0].str.strip()

# Apply normalization mapping
df['brand'] = df['brand'].replace(brand_mapping)

# Validation
unknown = df[~df['brand'].isin(known_brands)]['brand'].value_counts()
print('Unknown Brands:')
print(unknown if not unknown.empty else 'None.')

print(f'Brand Distribution: {df['brand'].value_counts()}')
```

### **2. Categorize Processor Brand**

Grouping processors into major brands: Intel, AMD, Snapdragon, MediaTek, Kirin, Apple.

```python
def categorize_processor(proc):
  if pd.isna(proc):
    return 'Other'
  p = proc.lower()
  if 'apple' in p:
    return 'Apple'
  if 'intel' in p:
    return 'Intel'
  if 'amd' in p:
    return 'AMD'
  if 'snapdragon' in p:
    return 'Snapdragon'
  if 'mediatek' in p:
    return 'MediaTek'
  if 'kirin' in p:
    return 'Kirin'
  return 'Other'

df['processor_brand'] = df['processor'].apply(categorize_processor)

print(f'Processor Brand Distribution: {df['processor_brand'].value_counts()}')
```

### **3. Categorize Price**

| Category | Price (IDR)        |
|----------|--------------------|
| Low      | < 10 Million       |
| Entry    |  10 – 15 Million   |
| Premium  |  15 – 30 Million   |
| Flagship |  > 30 Million      |

```python
def categorize_price(price):
  if pd.isna(price):
    return 'Unknown'
  if price < 10_000_000:
    return 'Low'
  if price < 15_000_000:
    return 'Entry'
  if price < 30_000_000:
    return 'Premium'
  return 'Flagship'

df['price_category'] = df['price_idr'].apply(categorize_price)

print(f'Price Category Distribution: {df['price_category'].value_counts()}')
```

### **4. Categorize Laptops**

Categories are determined based on the following **priority order**:

1. **MacBook** = Apple processor or name contains 'macbook'
2. **Chromebook** = name contains 'chromebook'
3. **Workstation** = Quadro/RTX A-series/RTX 2000/RTX 3000 GPU, or name contains ThinkPad P / ZBook / ProArt / MSI Creator
4. **Gaming** = GeForce RTX / Radeon RX / NVIDIA GPU
5. **2-in-1** = name contains Flex, Flip, Spectre, x360, ROG Flow, 2in1, Surface Pro, Spin
6. **Business** = name contains TravelMate, ExpertBook, Latitude, EliteBook, ProBook, ThinkPad
7. **Ultra** = processor ending in P/V, Snapdragon, Ryzen AI, 125H/225H/285H, or H-series without a gaming dGPU, with a price ≥ 15.000.000 IDR
8. **Daily** = MediaTek, Celeron, Pentium processor, or U-suffix with a price < IDR 15.000.000

```python
def categorize_laptop(row):
    name  = str(row['name']).lower()          if pd.notna(row['name'])          else ''
    proc  = str(row['processor']).lower()     if pd.notna(row['processor'])     else ''
    gpu   = str(row['graphics_card']).lower() if pd.notna(row['graphics_card']) else ''
    price = row['price_idr']                  if pd.notna(row['price_idr'])          else 0
    proc_raw = str(row['processor'])          if pd.notna(row['processor'])     else ''

    # 1. MacBook
    if 'apple' in proc or 'macbook' in name:
        return 'MacBook'

    # 2. Chromebook
    if 'chromebook' in name:
        return 'Chromebook'

    # 3. Workstation
    workstation_gpu  = ['quadro', 'a500', 'rtx 2000', 'rtx 3000', 'rtx a', 'rtx 500']
    workstation_name = ['thinkpad p', 'zbook', 'proart', 'msi creator']
    if any(k in gpu for k in workstation_gpu) or any(k in name for k in workstation_name):
        return 'Workstation'

    # 4. Gaming
    gaming_gpu = ['geforce rtx', 'nvidia rtx', 'radeon rx', 'nvidia geforce', 'rtx 4', 'rtx 3', 'rtx 5', 'rtx 2050']
    if any(k in gpu for k in gaming_gpu):
        return 'Gaming'

    # 5. 2-in-1
    kw_2in1 = ['flex', 'flip', 'spectre', 'x360', 'rog flow', '2in1', '2-in-1', 'surface pro', 'spin']
    if any(k in name for k in kw_2in1):
        return '2-in-1'

    # 6. Business
    business_kw = ['travelmate', 'expertbook', 'latitude', 'elitebook', 'probook', 'thinkpad']
    if any(k in name for k in business_kw):
        return 'Business'

    # 7. Ultrabook
    is_suffix_pv  = bool(re.search(r'\b\d+[PV]\b', proc_raw))
    is_snapdragon = 'snapdragon' in proc or 'qualcomm' in proc
    is_ryzen_ai   = 'ryzen ai' in proc
    is_h_special  = bool(re.search(r'\b(125H|155H|185H|225H|255H|285H)\b', proc_raw))
    is_h_no_dgpu  = bool(re.search(r'\d+HS?\b', proc_raw)) and \
                    not any(k in gpu for k in ['rtx', 'radeon rx', 'radeon fx'])
    if (is_suffix_pv or is_snapdragon or is_ryzen_ai or is_h_special or is_h_no_dgpu) and price >= 10_000_000:
        return 'Ultrabook'

    # 8. Daily
    daily_proc    = ['mediatek', 'celeron', 'pentium']
    is_suffix_u   = bool(re.search(r'\d+U\b', proc_raw))
    if any(k in proc for k in daily_proc) or \
       (is_suffix_u and price < 10_000_000) or \
       (is_snapdragon and price < 10_000_000):
        return 'Daily'

    return 'Daily'


df['laptop_category'] = df.apply(categorize_laptop, axis=1)

print(f'Price Category Distribution: {df['laptop_category'].value_counts()}')
```

### **5. Categorize RAM**

```python
ram_extracted = df['ram'].str.extract(r'(\d+)\s*GB', expand=False)
df['ram_gb'] = pd.to_numeric(ram_extracted, errors='coerce').astype('Int64')

print(df[['ram', 'ram_gb']].head(5))
print(f'NaN: {df["ram_gb"].isna().sum()}')
```

### **6. Categorize Storage**

```python
def parse_storage(val):
    val = str(val).strip()
    match_tb = re.search(r'(\d+(?:\.\d+)?)\s*TB', val, re.IGNORECASE)
    match_gb = re.search(r'(\d+)\s*GB', val, re.IGNORECASE)
    if match_tb:
        return int(float(match_tb.group(1)) * 1024)
    if match_gb:
        return int(match_gb.group(1))
    return pd.NA

df['storage_gb'] = df['storage'].apply(parse_storage)
df['storage_gb'] = pd.to_numeric(df['storage_gb'], errors='coerce').astype('Int64')

print(df[['storage', 'storage_gb']].head(5))
print(f'NaN: {df["storage_gb"].isna().sum()}')
```

### **7. Categorize GPU**

```python
def categorize_gpu(gpu):
    if pd.isna(gpu):
        return 'Integrated'
    g = gpu.lower()

    if any(k in g for k in ['rtx', 'gtx', 'quadro', 'nvidia']):
        return 'NVIDIA'
    if any(k in g for k in ['radeon rx', 'radeon 8', 'radeon 6', 'radeon vega']):
        return 'AMD Dedicated'
    if re.search(r'arc\s+a\d{3}', g):
        return 'Intel Arc Dedicated'
    if 'adreno' in g:
        return 'Qualcomm Adreno'
    if 'maleoon' in g:
        return 'Kirin Maleoon'
    if any(k in g for k in ['mali', 'immortalis']):
        return 'ARM Mali'
    if 'apple' in g or re.search(r'\d+-core gpu', g):
        return 'Apple GPU'
    return 'Integrated'

df['gpu_type'] = df['graphics_card'].apply(categorize_gpu)

print(df['gpu_type'].value_counts())
```

# **Modelling**

### **1. Combined text feature**

```python
df['combined_text'] = (
    df['brand']             + ' ' +
    df['processor_brand']   + ' ' +
    df['laptop_category']   + ' ' +
    df['price_category']    + ' ' +
    df['gpu_type']          + ' ' +
    df['os'].fillna('')     + ' ' +
    df['disk_type'].fillna('')
)
```

### **2. TF-IDF pada combined text**

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df['combined_text'])

print('TF-IDF matrix shape:', tfidf_matrix.shape)
```

### **3. Normalize Numerical Feature**

```python
num_cols = ['ram_gb', 'storage_gb', 'price_idr', 'rating']

df_num = df[num_cols].copy()
df_num['ram_gb'] = df_num['ram_gb'].fillna(df_num['ram_gb'].median())
df_num['storage_gb'] = df_num['storage_gb'].fillna(df_num['storage_gb'].median())

scaler = MinMaxScaler()
num_scaled = scaler.fit_transform(df_num)

print('Numerical matrix shape:', num_scaled.shape)
```

### **4. Combine TF-IDF + numerik = Cosine Similarity**

```python
weight_text = 0.6
weight_num = 0.4

num_sparse = sp.csr_matrix(num_scaled * weight_num)
tfidf_weighted = tfidf_matrix * weight_text

combined_matrix = sp.hstack([tfidf_weighted, num_sparse])
cosine_sim = cosine_similarity(combined_matrix, combined_matrix)

print('Combined matrix shape:', combined_matrix.shape)
```

### **5. Recommendation**

```python
df = df.reset_index(drop=True)
indices = pd.Series(df.index, index=df['name'])

laptop_input = 'Asus Gaming V16 2025 V3607VH-RP038WS Laptop'
top_n = 10

idx = indices[laptop_input]
scores = list(enumerate(cosine_sim[idx]))
scores = sorted(scores, key=lambda x: x[1], reverse=True)
scores = scores[1:top_n+1]

result_idx = [i[0] for i in scores]

cols_show = ['name', 'processor', 'graphics_card', 'ram_gb', 'screen_size', 'storage', 'os',
             'price_idr', 'rating', 'laptop_category']

print(f'Rekomendasi untuk: {laptop_input}\n')
df.iloc[result_idx][cols_show]
```
